```python
%%markdown
# Week 2 Project 
Temperature Converter
Celsius to Fahrenheit and Fahrenheit to Celsius.
```

```python
%%markdown
# Formulas 
🟡Fahrenheit = (Celsius * 9 / 5) + 32 
🟡Celsius = (Fahrenheit - 32) * 5 / 9
```

```python
print ("========== Temperature Converter ==========")
# Celsius to Fahrenheit 
celsius = float(input("Enter Temperature In Celsius :"))
fahrenheit_ans =  (celsius * 9 / 5) + 32
print (celsius , "Celsius Is Equal To",fahrenheit_ans,"Fahrenheit")

# Fahrenheit to Celsius 
fahrenheit = float(input("Enter Temperature In Fahrenheit :"))
celsius_ans = (fahrenheit - 32) * 5 / 9
print (fahrenheit , "Fahrenheit Is Equal To",celsius_ans ,"Celsius")
```
